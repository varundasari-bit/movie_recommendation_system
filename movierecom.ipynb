```python
# Store null rows
null_rows = movies[movies.isnull().any(axis=1)]

# Remove null rows
movies_cleaned = movies.dropna()

# Store duplicate rows
duplicate_rows = movies_cleaned[
    movies_cleaned.duplicated(subset="title", keep="first")
]

# Remove duplicate rows
movies_cleaned = movies_cleaned.drop_duplicates(subset="title")

# Reset index
movies_cleaned = movies_cleaned.reset_index(drop=True)
```

```python
import ast
def names(text):
    if isinstance(text, list):
        return text    # If already a list, return it
    return [d["name"] for d in ast.literal_eval(text)]
def director(text):
    if isinstance(text, list):
        return text    # If already a list of names, return it
    return [
        d["name"]
        for d in ast.literal_eval(text)
        if d["job"] == "Director"
    ]
movies["genres"] = movies["genres"].apply(names)
movies["keywords"] = movies["keywords"].apply(names)
movies["cast"] = movies["cast"].apply(lambda x: names(x)[:3])
movies["crew"] = movies["crew"].apply(director)
movies.iloc[0]
```

```python
def overview(text):
  if isinstance(text,list):
    return text
  return text.split()
movies["overview"] = movies["overview"].apply(overview)

for col in ["genres", "keywords", "cast", "crew"]:
    movies[col] = movies[col].apply(lambda x: [i.replace(" ", "") for i in x])

movies["tags"] = (
    movies["overview"]
    + movies["genres"]
    + movies["keywords"]
    + movies["cast"]
    + movies["crew"]
)

```

```python
print(type(movies["overview"]))

print(movies.columns[movies.columns.duplicated()])

print(movies.loc[:, movies.columns == "overview"].head())
```

```python
print(movies["tags"].head())
print(type(movies["tags"].iloc[0]))
```

```python
movies["tags"] = movies["tags"].apply(
    lambda x: " ".join(x)
)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words="english")
vectors=cv.fit_transform(movies ["tags"]).toarray()
print(vectors.shape)
```

```python
# Number of movies
print("Movies:", len(movies))

# Number of tags
print("Tags:", len(movies["tags"]))

# Check missing tags
print("Missing tags:", movies["tags"].isna().sum())

# Check empty tags
print("Empty tags:", (movies["tags"].str.strip() == "").sum())

# Check vector dimensions
print("Vector shape:", vectors.shape)

# Check vocabulary
print(cv.get_feature_names_out()[:30])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)

print(similarity.shape)
```

```python
print(similarity.shape)

# Similarity of first movie with itself
print(similarity[0][0])

# Similarity of first movie with second movie
print(similarity[0][1])

# Similarity of first movie with third movie
print(similarity[0][2])

# Highest and lowest similarity
print(similarity.max())
print(similarity.min())
```

```python
print(similarity[0][:10])
```

```python
def recommend(movie):

    index = movies[movies["title"] == movie].index[0]

    scores = list(enumerate(similarity[index]))

    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    for i, score in scores[1:6]:
        print(movies.iloc[i].title)
```

```python
movie_name = input("Enter a movie name: ")
recommend(movie_name)
```
